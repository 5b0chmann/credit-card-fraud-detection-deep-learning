```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

results = pd.DataFrame ({"Model":["Logistic Regression",
                                "Random Forest", 
                                "Deep Neural Network",
                                "Autoencoder"],
                        "Precision": [0.66, 0.96, 0.89, 0.15], "Recall":[0.79, 0.73, 0.77, 0.58],
                        "F1-Score": [0.72, 0.83, 0.82, 0.24], "ROC-AUC": [0.9657, 0.9663, 0.9642, 0.9145]})

# Ranking-Tabelle
ranking = results.sort_values(by = "F1-Score", ascending = False)
print("Model Ranking by F1 Score")
print(ranking[["Model", "F1-Score"]])

# Precision Comparison
plt.figure(figsize=(8, 5))

ax = sns.barplot(data=results, x="Model", y="Precision")
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")
plt.title("Precision Comparison")
plt.xticks(rotation=20)
plt.show()

# Recall Comparision
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=results, x="Model", y="Recall")
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")
plt.title("Recall Comparison")
plt.xticks(rotation=20)
plt.show()

# F1 Score Comparision
plt.figure(figsize = (8, 5))
ax = sns.barplot(data = results, x = "Model", y = "F1-Score")
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")
plt.title("F1 Score Comparison")
plt.xticks(rotation= 20)
plt.show()

# ROC-AUC Comparison
plt.figure(figsize = (8, 5))
ax = sns.barplot(data = results, x = "Model", y = "ROC-AUC")
for container in ax.containers:
    ax.bar_label(container, fmt="%.4f")
plt.title("ROC-AUC Comparison")
plt.xticks(rotation = 20)
plt.show()

# combines Visualization
metrics = results.melt(id_vars = "Model",
                       var_name = "Metric",
                       value_name = "Score")
plt.figure(figsize = (12, 6))
sns.barplot(data = metrics, x = "Metric",
            y = "Score",
            hue = "Model")

plt.title("Model Performance Comparison")
plt.legend(bbox_to_anchor = (1.15, 1), loc = "upper left")
plt.show()
```

Although Logistic Regression, Random Forest and the Deep Neural Network achieved very similar ROC-AUC scores, substantial differences were observed in Precision and F1-Score. This demonstrates that ROC-AUC alone is not sufficient for evaluating highly imbalanced fraud detection datasets.
